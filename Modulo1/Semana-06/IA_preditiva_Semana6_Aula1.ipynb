## Pandas exportar dataframes

```python
import pandas as pd

df_dados = pd.DataFrame({
    'nome': ['Ana', 'Bruno', 'Carla'],
    'idade': [25, 30, 35],
    'cidade': ['SP', 'RJ', 'MG']
})

# Componentes principais
print("Índices:", list(df_dados.index))
# Índices: RangeIndex(start=0, stop=3, step=1)

print("Colunas:", list(df_dados.columns))
# Colunas: Index(['nome', 'idade', 'cidade'], dtype='object')

print("Dados:\n", df_dados) 
```

```python
# exportando em .csv
df_dados.to_csv('dados_csv.csv')
df_dados.to_csv('dados_csv_sem_indice.csv', index=False)
df_dados.to_csv('dados_csv_cidades.csv', index=False, columns=['nome', 'cidade'])
```

```python
# exportando em .xlsx
# %pip install openpyxl
df_dados.to_excel('dados_excel.xlsx', index=False)

df1 = pd.DataFrame({
    'Produto': ["TV", "Celular"],
    'Preco': [1200, 2500]
    }
)
df2 = pd.DataFrame({
    'Nome': ["Nicodemus", "Esequiel"],
    'Cidade': ["SP", "SC"] 
    }
)

with pd.ExcelWriter('multi_abas_excel.xlsx') as writer:
    df1.to_excel(writer, sheet_name='Vendas')
    df2.to_excel(writer, sheet_name='Clientes')
```

```python
# exportando em .json

df = pd.DataFrame({
    'nome': ['Ana', 'Bruno'],
    'idade': [25, 30]
})

# JSON index
df.to_json('dados_json.json', orient='index') 

# JSON records
df.to_json('dados_padrao.json', orient='records', force_ascii=False)

```

```python
# exportando em .txt

# Tab separado
df_dados.to_csv('dados_txt.txt', sep='\t', index=False)

# XML (pandas >= 1.0)
# %pip install lxml 
df.to_xml('dados_xml.xml', index=False) 

```
