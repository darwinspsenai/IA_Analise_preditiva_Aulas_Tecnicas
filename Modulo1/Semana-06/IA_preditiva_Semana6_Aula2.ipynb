## Datatime

```python
from datetime import date
#import datetime

hoje = date.today()
print("Data atual:", hoje, "tipo:", type(hoje), "Dia:", hoje.day)

# Componentes
print("Ano:", hoje.year, "Mes:", hoje.month, "Dia:", hoje.day)

```

```python
from datetime import time

h = time(14, 30, 45)
print("Horário:", h, "Hora:", h.hour)

# Componentes
print("Hora:", h.hour, "Minuto:", h.minute, "Segundo:", h.second)

# Com microsegundos
t = time(14, 30, 45, 123456)
print("Preciso:", t)
```

```python
from datetime import datetime

dt = datetime(year=2024, month=12, day=25, hour=10, minute=30, second=15)
print("DateTime:", dt, "Data:", dt.date(), "Tipo:", type(dt))

print("Ano:", dt.year, "Mes:", dt.month, "Dia:", dt.day)
print("Hora:", dt.hour, "Minuto:", dt.minute, "Segundo:", dt.second)

agora = datetime.now()
print("Ano:", agora.year, "Mes:", agora.month, "Dia:", agora.day)
print("Hora:", agora.hour, "Minuto:", agora.minute, "Segundo:", agora.second)
```

```python
agora = datetime.now()
print(agora.strftime("%d/%m/%Y"))

print(agora.strftime("%H:%M:%S"))

# diferença
d1 = date(year=2026, month=10, day=9)
d2 = date(2026, 11, 23)
diff = d2 - d1
print("Dias:", diff.days)
# Dias: 365 

```

```python
# datetime em pandas
import pandas as pd

df = pd.DataFrame({
    'data': ['2024-01-01', '2024-02-15', '2024-03-20']
})

print(df['data'].dtype)
df['data'] = pd.to_datetime(df['data'])
print(df['data'].dtype)

print(df)

df['ano'] = df['data'].dt.year
df['mes'] = df['data'].dt.month
df['dia'] = df['data'].dt.day

print(df)
```

```python
agora = datetime.now()

# Formatos comuns
print("ISO:", agora.strftime("%Y-%m-%d"))
# ISO: 2024-10-06

print("BR:", agora.strftime("%d/%m/%Y"))
# BR: 06/10/2024

```

```python
# importar json em pandas

df_json = pd.read_json('dados_padrao.json')
print(df_json)

# APIs
#%pip install requests
import requests

cep = "13083650"
url = f"https://viacep.com.br/ws/{cep}/json/"

try:
    # Fetch the online JSON data
    response = requests.get(url)

    if response.status_code == 200:
        df = pd.json_normalize(response.json())
        print(df)
    else:
        print("Status code: ", response.status_code)
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
```
