```python
class Carro:
    categoria = "transporte" # self por padrão

    def __init__(self, marca, cor, potencia=None):
        self.marca = marca
        self.cor = cor
        self.potencia = potencia
        self.velocidade = 0

    def apresentar(self):
        "marca: str"
        print('Carro da marca ' + self.marca + ' com a cor ' + self.cor\
              + ' com aceleracao de ' + str(self.velocidade) )

    def pintar(self, nova_cor):
        "nova_cor: str"
        self.cor = nova_cor

    def acelerar(self, valor):
        self.velocidade += valor

    def mudar_categoria(self, nova_categoria):
        self.categoria = nova_categoria



carro1 = Carro("Fiat", "branca")
carro2 = Carro("Sandero", "preta")
carro2.cor = "vermelho" # nao boa pratica
carro2.acelerar(37)
carro3 = Carro(cor='prata', marca='Honda')


#print(carro1.marca, carro1.cor, carro1.potencia)
#print(carro2.marca, carro2.cor)
flota_carros = [carro1, carro2, carro3]

for instancia_carro in flota_carros:
    print(vars(instancia_carro))
    #print(instancia_carro.marca, instancia_carro.cor)
    instancia_carro.apresentar()
    print("---------------")

carro1.apresentar() # chamando um metodo de uma instancia/objeto

Carro.apresentar(carro2) # chamando ao metodo apresentar e 
                         # passando o objeto carro2 como parametro
```

```python
class Conta:
    def __init__(self, titular):
        self.titular = titular
        self.saldo = 0.0

    def apresentar(self):
        print("titulo:", self.titular, "Saldo:", self.saldo)

    def adicionar_saldo(self, valor_adicionar):
        "valor_adicionar: float"
        self.saldo = self.saldo + valor_adicionar

    @property
    def func_saldo(self):
        return self.saldo

conta1 = Conta("Lucas")
conta1.apresentar()
print( "Saldo de conta1:", conta1.func_saldo )

conta2 = Conta("Ana")
conta2.adicionar_saldo(58.6)
print( "Saldo de conta2:", conta2.func_saldo )
```

```python
class Motor():
    def __init__(self, potencia):
        self.potencia = potencia

    def mostrar(self):
        print("potencia:", self.potencia)


class Pneu():
    def __init__(self, diametro_roda, indice_carga):
        self.roda = diametro_roda
        self.carga = indice_carga

    def mostrar(self):
        print("roda:", self.roda, "carga:", self.carga)

    def atualizar_pneu(self, roda, carga):
        "roda: int"
        "carga: int"
        self.carga = carga
        self.roda = roda


class AutoMovel():
    pneu = Pneu(18, 89)

    def __init__(self, param_marca, param_motor):
        self.marca = param_marca
        self.motor = param_motor

    def trocar_pneu(self, novo_pneu):
        "novo_pneu: class Pneu"
        self.pneu = novo_pneu

    def trocar_pneu2(self, roda, carga):
        "roda: int"
        "carga: int"
        pneu.atualizar_pneu(roda, carga)


motor = Motor(1.6)
camionete = AutoMovel(param_marca = "Toyota", param_motor = motor)

# chamar ao metodo mostrar da classe motor
camionete.motor.mostrar()

# chamar ao metodo mostrar da classe pneu
# usar o metodo troca_pneu passado o parametro pneu(objeto)
pneu = Pneu(diametro_roda = 16, indice_carga = 91)
camionete.trocar_pneu(pneu)
camionete.pneu.mostrar()

# usar o metodo trocar_penu2 pasando os parametros roda e carga
camionete.trocar_pneu2(roda = 14, carga = 87)
camionete.pneu.mostrar()
```

```python
class VeiculoBase:
    cor = None

    def __init__(self, qnt_rodas, cor):
        "qnt_rodas: int"
        self.qnt_rodas = qnt_rodas
        self.cor = cor

    def definir_cor(self, cor):
        "cor: str"
        self.cor = cor

    def definir_qnt_rodas(self, qnt_rodas):
        "qnt_rodas: int"
        self.qnt_rodas = qnt_rodas

    def mostrar(self):
        print("cor:", self.cor, "qnt_rodas:", self.qnt_rodas)


class Bicicleta(VeiculoBase):

    def __init__(self, aro, qnt_rodas):
        self.aro = aro
        self.definir_qnt_rodas(qnt_rodas)

    def trocar_aro(self, num_aro):
        "num_aro: int"
        self.aro = num_aro

class Camiao(VeiculoBase):

    def __init__(self, qnt_rodas, cor, potencia_motor):
        "qnt_rodas: int"
        "potencia_motor: float "
        #self.definir_qnt_rodas(qnt_rodas)
        super().__init__(qnt_rodas, cor)
        self.potencia_motor = potencia_motor
        

    def atualizar_potencia(self, potencia):
        "potencia: float"
        self.potencia_motor = potencia
    
    
bike = Bicicleta(aro = 26, qnt_rodas = 2)
bike.definir_cor("verde")
bike.mostrar()

camiao = Camiao(qnt_rodas=4, cor="roxo", potencia_motor=2.0)
#camiao.definir_cor("azu")
camiao.mostrar()
```
