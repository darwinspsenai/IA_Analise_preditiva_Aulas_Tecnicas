### Exercicio

```python
'''
Uma família deseja analisar o consumo de energia elétrica durante 7 dias.
Para cada dia, é informado o consumo de energia em kWh.
O programa deve informar:
O consumo total da semana;
O consumo médio diário;
O dia de maior consumo;
Quantos dias tiveram consumo acima da média;
Quantos dias tiveram consumo menor ou igual a 10 kWh.

Entrada
8 12 15 9 20 11 7

Saída
Consumo total: 82
Consumo medio: 11.714285714285714
Dia de maior consumo: 5
Dias acima da media: 3
Dias com consumo menor ou igual a 10: 3
'''

kwh_semana = [8, 12, 15, 9, 20, 11, 7]

soma_kwh = 0
consumo_meio = 0
kwh_maior = 0
dia_maior = 0
dias_menor_igual = 0
id = 0

while id < len(kwh_semana):
    # soma kwn da semana
    soma_kwh += kwh_semana[id] 

    # atualizar kwh_maior
    if kwh_maior < kwh_semana[id]:
        kwh_maior = kwh_semana[id]
        dia_maior = id

    # Dias com consumo menor ou igual a 10
    if kwh_semana[id] <= 10:
        dias_menor_igual += 1 

    id += 1

# meia aritmetica da semana
consumo_meio = soma_kwh / 7 # len(kwh_semana)

idx = 0
contador_maior_meia = 0

while idx < len(kwh_semana):
    # dias maiores do que a meia aritmetica
    if kwh_semana[idx] > consumo_meio: 
        contador_maior_meia += 1

    idx += 1


print("Consumo total:", soma_kwh)
print("Consumo medio:", consumo_meio)
print("Dia de maior consumo:", dia_maior+1)
print("Dias acima da media: ", contador_maior_meia)
print("Dias com consumo menor ou igual a 10:", dias_menor_igual)
```

### Pandas

```python
import pandas as pd

dados = {
   "Nome": ["Ana", "Bruno", "Carla"],
   "Idade": [21, 20, 22] 
}
print(dados)

ind = ["A", "B", "C"]
df_dados = pd.DataFrame(data=dados, index=ind)
print(df_dados)

df_dados["Sexo"] = ["F", "M", "F"]
df_dados["Idade"] = [34, 32, 37]
print(df_dados)

print("--------------")
dados_2 = {
    "Nome":["Antonio", "Isabela"], 
    "Idade": [50, 22],
    "Sexo": ["M", "F"]
}

df_dados_2 = pd.DataFrame(data=dados_2)

df_concat = pd.concat([df_dados, df_dados_2], ignore_index=True)

print(df_concat)
```

```python
print(df_concat)
print(df_concat.iloc[3])
df_concat.iloc[3] = ["Marcos", 45, "M"]
print(df_concat)
print("-----------")
df_concat.index = ["A", "B", "C", "D", "E"]

print(df_concat)
df_concat.loc["B"] = ["Ademir", 25, "M"]
print(df_concat)
```

```python
print(df_concat)
df_newconcat = df_concat.copy(deep=True)
df_newconcat.at["E", "Idade"] = 20
df_newconcat.iat[2, 0] = "Maria"
print(df_newconcat)
```

```python
print(df_concat)
df_newconcat_3 = df_concat.copy(deep=True)
df_newconcat_3.drop(index=['B', 'D'], columns=['Idade'], inplace=True)
print(df_newconcat_3)
```
