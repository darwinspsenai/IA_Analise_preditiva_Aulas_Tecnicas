```python
"""
Exercício 2: Análise de vendas da semana
Problema
Uma pequena loja quer analisar suas vendas durante uma semana.
Para cada dia, o vendedor informar o valor total vendido naquele dia.
Você deve calcular:
O total vendido durante a semana;
A média de vendas por dia;
Qual foi o dia com maior venda;
Quantos dias tiveram vendas acima da média.

Considere:
Segunda-feira = dia 1
Terça-feira = dia 2
...
Domingo = dia 7

Entrada
Serão informados 7 valores, um para cada dia da semana.

Saída
Imprima:
o total vendido;
a média;
o dia de maior venda;
a quantidade de dias acima da média.
"""

# Entrada
# 100 200 150 300 250 180 220

# Saida 
# Total: 1400 
# Media: 200.0 
# Melhor dia: 4 
# Dias acima da media: 3 

total_soma = 0
media_aritmetica = 0 # ∑ x_i / n
id_melhor = 0
maximo_atual = 0
contador_maior_media = 0

valor_semana = [100, 200, 150, 300, 250, 180, 220] # [seg, ter, quart, quint, sex, sab, dom]

for i in range(len(valor_semana)): #[0, 1, 2, 3, 4, 5, 6]
    # soma das vendas
    total_soma += valor_semana[i]

    # conferindo a maior venda
    if maximo_atual < valor_semana[i]:
        maximo_atual = valor_semana[i]
        id_melhor = i

media_aritmetica = total_soma / len(valor_semana)

for i in range(len(valor_semana)):
    # condicao se e maior que a media aritmetica
    if valor_semana[i] > media_aritmetica:
        contador_maior_media += 1

print("Total: ", total_soma)
print("Media: ", media_aritmetica)
print("Melhor dia: ", id_melhor+1, " maximo venda e: ", maximo_atual)
print("Dias acima da media: ", contador_maior_media)
```

```python
# criando nosso primeiro dataframe
list_nomes = ["Carla", "Jose", "Artur", "Rosaly"]
list_idades = [ 24, 37, 33, 18]
list_nome_coluna = ["nomes", "idades"]
list_idx = ["pessoa1", "pessoa2", "pessoa3", "pessoa4"]

dados = list(zip(list_nomes, list_idades))
print(dados)

df_dados = pd.DataFrame(data = dados, columns = list_nome_coluna, index = list_idx) #columns = ["nomes", "idades"])
print(df_dados)
```
