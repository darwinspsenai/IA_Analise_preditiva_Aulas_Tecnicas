```python
"""
Exercício 1: Controle de fila no mercado 

Problema
Um supermercado possui uma fila de clientes. Para cada cliente, é informado o número de produtos que ele possui no carrinho.
O caixa consegue atender um produto por segundo.
Você deve determinar:
Quanto tempo será necessário para atender todos os clientes;
Qual cliente possui o maior número de produtos;
Quantos clientes possuem 10 produtos ou mais.

Entrada
Uma linha contém números inteiros, representando a quantidade de produtos de cada cliente.

Saída
Imprima:
o tempo total de atendimento em segundos;
o número do cliente com mais produtos;
a quantidade de clientes com 10 ou mais produtos.

Considere que o primeiro cliente é o cliente 1.

Entrada
4 12 7 15 3

Saida:
Tempo total: 41 
Cliente com mais produtos: 4 
Clientes com 10 ou mais produtos: 2
"""

lista_cliente = [4, 12, 7, 15, 3]

tempo_total = 0
id_cliente_mais_produtos = 0
max_qnd_produto = lista_cliente[0]
cliente_mais_10_prod = 0

for id in range(len(lista_cliente)): #[0, 1, 2, 3, 4]
    # contagem de tempo total de caixa
    tempo_total = tempo_total + lista_cliente[id]
    # tempo_total += lista_cliente[id]

    # cliente com qnd de produto maximo
    if max_qnd_produto < lista_cliente[id]:
        max_qnd_produto = lista_cliente[id]
        id_cliente_mais_produtos = id

    # clientes com mais de 10 produtos
    if lista_cliente[id] >= 10:
        cliente_mais_10_prod += 1

print("Tempo total:", tempo_total)
print("Cliente com mais produtos:", id_cliente_mais_produtos+1, "com qnd:", max_qnd_produto)
print("Clientes com 10 ou mais produtos:", cliente_mais_10_prod)
```

```python
"""
Entrada
  2 3 4 5 

Saida 
  Tempo total: 14 
  Cliente com mais produtos: 4 
  Clientes com 10 ou mais produtos: 0
""" 

lista_cliente = [2, 3, 4, 5]

tempo_total = 0
id_cliente_mais_produtos = 0
max_qnd_produto = lista_cliente[0]
cliente_mais_10_prod = 0
id = 0

while id < len(lista_cliente):
  # contagem de tempo total de caixa
  tempo_total = tempo_total + lista_cliente[id]

  # cliente com qnd de produto maximo
  if max_qnd_produto < lista_cliente[id]:
    max_qnd_produto = lista_cliente[id]
    id_cliente_mais_produtos = id

  # clientes com mais de 10 produtos
  if lista_cliente[id] >= 10:
    cliente_mais_10_prod += 1

  id += 1

print("Tempo total:", tempo_total)
print("Cliente com mais produtos:", id_cliente_mais_produtos+1, "com qnd:", max_qnd_produto)
print("Clientes com 10 ou mais produtos:", cliente_mais_10_prod)
```

```python
# exercicio
# dia/mes/ano: Temperatura: temp
dados = [
    { # data1
        "dia":"12", 
        "mes":10, 
        "ano":2021, 
        "temp": 30.5
    },
    { # data2
        "dia":25, 
        "mes":2, 
        "ano":2026, 
        "temp": 19.8
    },
]

for var_dict in dados:
    mes = var_dict["mes"]
    if mes < 10: # 1 ... 9
        mes = "0" + str(mes)
    else: # 11, 12
        mes = str(mes)

    linha = str(var_dict["dia"]) + "/" + mes \
        + "/" + str(var_dict["ano"]) + " Temperatura: " + str(var_dict["temp"])
    print(type(linha), linha)
```
