```python
npprodutos = np.array(produtos)
npprecos = np.array([10.53, 3.21, 5.47])

# para lista mas array tbm faz
receitas = np.zeros(len(produtos))
for i in range(len(produtos)):
    receitas[i] = quantidades[i] * npprecos[i]

print("produtos: ", type(npprodutos), npprodutos)
print("quantidade: ", type(quantidades), quantidades)
print("preco: ", type(npprecos), npprecos)
print("receita: ", type(receitas), receitas)

# operação vectorial.
#            [20 35 15] * [10.53  3.21  5.47]
npreceitas = quantidades * npprecos # [210.6  112.35  82.05]
print("npreceitas: ", type(npreceitas), npreceitas)
```

```python
# Calcula preços após um desconto de 11%.
precos_desconto_11 = npprecos * 0.89
print("precos: ", npprecos)
print("precos desconto: ", precos_desconto_11)

precos_desconto_python = [0] * 3
for i in range(len(precos_desconto_11)):
    precos_desconto_python[i] = round(precos_desconto_11[i], 2)
print("precos_desconto_python: ", precos_desconto_python)

# Apresenta os valores com até duas casas decimais.
precos_arredondados = np.round(precos_desconto_11, 2)
print("precos_arredondados:", precos_arredondados)
```

```python
# Calcula a receita total do primeiro período.
receita_total = receitas.sum() # np.sum(receitas, axis=0)

# Divide cada receita pelo mesmo total.
#        ( [210.6  112.35  82.05] / 405.0 ) * 100
participacao_receita = (receitas / receita_total) * 100

print("Receitas: ", receitas)
print("Receita total:", receita_total)
print("Participações (%):", np.round(participacao_receita, 2))
```

```python
# Cada função recebe o array que será resumido.
print("quantidades: ", quantidades)
print("Total de unidades:", np.sum(quantidades))
print("Média por produto:", np.round(quantidades.mean(), 2))
print("Menor quantidade:", np.min(quantidades))
print("Maior quantidade:", np.max(quantidades))
print("Desvio padrão quantidade:", np.round(np.std(quantidades), 1))
```

```python
# Verifica quais produtos venderam pelo menos 20 unidades.
mascara_20 = quantidades >= 20

print(quantidades)
print(mascara_20)

# filtrando só as qnt maiores ou iguais do que 20
#quantidades_maior_20 = quantidades[mascara_20]
quantidades_maior_20 = quantidades[quantidades >= 20]
print(quantidades_maior_20)
```

```python
# As duas condições precisam ser verdadeiras.
#                   [ True  True False] & [ True  False True]
mascara_intervalo = (quantidades >= 20) & (quantidades <= 30)
quantidades_intervalo = quantidades[mascara_intervalo]

print("quantidades: ", quantidades)
print("filtro: ", mascara_intervalo)
print("quantidades filtrada: ", quantidades_intervalo, \
        type(quantidades_intervalo), quantidades_intervalo.shape)
```

```python
def np_barras_vertical(array, categoria, titulo=None):
    '''
    Param:
        tabela: dataframe
        categoria: str
        valor: str
        titulo: str
    '''
    fig, ax = plt.subplots()
    sns.barplot(x=categoria, y=array, hue=categoria,
                legend=False, errorbar=None,
                orient="v", width=0.72, ax=ax)
    ax.set_title(titulo)
    ax.grid(axis="x", visible=False)
    ax.margins(y=0.18)
    ax.tick_params(axis="x", labelrotation=28)
    for grupo in ax.containers:
        ax.bar_label(grupo, fmt="%.0f", padding=4, fontsize=10)
    sns.despine(ax=ax, left=True, bottom=True) # remover bordas
    fig.tight_layout() # ajusta automaticamente os espaçoes
    plt.show() 
```

```python
np_barras_vertical(receitas, npprodutos)
```
