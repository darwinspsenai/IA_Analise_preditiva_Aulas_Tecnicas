```python
# encontrar o maior numero
var_numeros = [12, 12, 5, 7, 1, 10]

indice = 0
num_max = var_numeros[0]
while indice < len(var_numeros):
    if var_numeros[indice] > num_max:
        num_max = var_numeros[indice]
    indice += 1

print("o numero maximo e: ", num_max)
```

```python
lista_nomes = ["pedro", 'lucas', "juan", "artur", "timoteo"]
indice = 0
while indice < len(lista_nomes):
    print(lista_nomes[indice])
    indice += 1
print("--------------------")
for valor_lista in lista_nomes:
    print(valor_lista)
```

```python
# encontrar o maior numero
var_numeros = [3, 13, 5, 17, 10]

maxi = var_numeros[0]
for valor in var_numeros:
    if valor > maxi:
        maxi = valor

print("numero maximo e ", maxi)
```
