```python
# dado 3 numeros imprimir em orden descente
A = 10
B = 7
C = 1

if A <= B:
    if B <= C:
        print(A, B, C) # quando A e o menor
    else: 
        if A <= C:
            print(A, C, B)
        else:
            print(C, A, B)
else: #a > b
    if A <= C:
        print(B, A, C)
    else:
        if B <= C:
            print(B, C, A)
        else:
            print(C, B, A)
```

```python
# B elevado E = B x B x B ... B
B = float(input())
E = int(input())

contador = 1
valor_final = 1
while contador <= E:
    valor_final = valor_final * B # valor_final *= B
    contador = contador + 1 # contador += 1

print("valor final: ", valor_final)
```

```python
# Faça um laço de repetição que printe os número de 1 até 20.
for valor in range(1, 21): # (inicio, final, passo=1)
    print(valor, end=" ")
```

```python
# Faça um programa que print na tela a mensagem “olá mundo” 10 vezes.
for i in range(10): # [0, 1, ... 9] = tamanho = 10
    print(i+1, "ola mundo")

contador = 0
while contador < 10:
    print(contador+1, "ola brasil")
    contador += 1
```

```python
# Faça um laço de repetição que printe os números de 0 até 100, porém pulando de dois em dois.

# range(inicio, final, passo)
for valor in range(0, 101, 2):
    print(valor, end="--")

print("") # adicionar um salto de linha
contador = 0 # inicio
while contador < 101: # final vai ser a condição
    print(contador, end="~~")
    contador += 2 # passo

```

```python
# Faça um laço de repetição que irá rodar 5 vezes
# a cada volta ele terá que usar um input para pedir um número 
# mas somente printar em tela os números maiores que 10

for i in range(5): # [0, 1, 2, 3, 4] # repetir 5 vezes
    numero = float(input("ingressar numero: ")) # pedir um numero
    if numero > 10: # maior do que 10
        print(numero)
print("final do for")

contador = 0
while contador < 5: # repetir 5 vezes
    numero = float(input("ingressar numero: ")) # pedir um numero
    if numero > 10: # maior do que 10
        print(numero)
    contador += 1
print("final do while")
```
