```python
# ingressar un numero > 1 e imprimir seus divisores
numero = int(input("ingressar numero: "))
print("numero: ", numero)
divisor = 1
while divisor < numero:
    if numero % divisor == 0:
        print(divisor, end=" ")
    divisor += 1
```

```python
# ingressar un numero > 1 e imprimir o numoro dos seus divisores
numero = int(input("ingressar numero: "))
print("numero: ", numero)
divisor = 1
contador_duvisores = 0
while divisor < numero:
    if numero % divisor == 0:
        contador_duvisores += 1
    divisor += 1
print("o numero de divisores e: ", contador_duvisores)
```
